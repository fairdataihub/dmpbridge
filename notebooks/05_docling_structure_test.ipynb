```python
from pathlib import Path

from dmpbridge.pdf.docling_extractor import save_docling_outputs
from dmpbridge.pdf.docling_postprocessor import save_docling_structured_blocks
from dmpbridge.processing.structure_json_builder import save_narrative_json
```

```python
project_root = Path.cwd().parent

pdf_path = project_root / "data" / "raw_pdfs" / "sample10.pdf"
skeleton_path = project_root / "schemas" / "rda_dmp_dmptool_extension_skeleton.json"

markdown_path = project_root / "data" / "docling_markdown" / f"{pdf_path.stem}.md"
structured_path = project_root / "data" / "docling_json" / f"{pdf_path.stem}_structured_blocks.json"
final_json_path = project_root / "data" / "structure_json" / f"{pdf_path.stem}_docling.json"

print("PDF exists:", pdf_path.exists())
print("Skeleton exists:", skeleton_path.exists())
```

```python
docling_result = save_docling_outputs(pdf_path)

print("Markdown saved:", markdown_path.exists())
print(docling_result["markdown"][:2000])
```

```python
docling_blocks = save_docling_structured_blocks(
    markdown_path=markdown_path,
    output_path=structured_path,
    source_pdf=pdf_path.name
)

print("Structured blocks:", len(docling_blocks))

for block in docling_blocks[:30]:
    print(block["label"], "→", block["text"])
```

```python
docling_json = save_narrative_json(
    structured_blocks=docling_blocks,
    output_path=final_json_path,
    skeleton_path=skeleton_path
)

print("Saved Docling JSON:", final_json_path.exists())
```

```python
sections = docling_json["narrative"]["template"]["section"]

print("Number of sections:", len(sections))

for section in sections:
    print(section["order"], section["title"], "| questions:", len(section["question"]))
```
