```python
from pathlib import Path

from dmpbridge.pdf.page_image_converter import convert_pdf_to_images
```

```python
project_root = Path.cwd().parent

pdf_path = project_root / "data" / "raw_pdfs" / "sample10.pdf"

print("Project root:", project_root)
print("PDF path:", pdf_path)
print("PDF exists:", pdf_path.exists())
```

```python
image_paths = convert_pdf_to_images(pdf_path, dpi=200)

print("Number of page images:", len(image_paths))

for image_path in image_paths:
    print(image_path)
    print("Exists:", image_path.exists())
```

```python
output_dir = project_root / "data" / "page_images" / pdf_path.stem

print("Output folder:", output_dir)
print("Output folder exists:", output_dir.exists())

list(output_dir.glob("*.png"))
```
