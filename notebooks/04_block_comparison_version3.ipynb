# DMPBridge Block Comparison Notebook — Order-Aware Evaluation

This notebook compares manual/reference JSON blocks with AI/Llama detected JSON blocks.

It evaluates:

- text similarity
- label correctness
- document order difference
- per-label accuracy
- confusion matrix
- separate performance for `document_title`, `section`, `subsection`, and `content`
- missing reference blocks
- extra AI-detected blocks

## Step 4 — Helper functions

```python
def normalize_text(text):
    """Normalize text for fair comparison while preserving original text in reports."""
    if text is None:
        return ""

    text = str(text)
    text = text.replace("\n", " ")
    text = text.replace("\t", " ")
    text = text.replace("“", '"').replace("”", '"')
    text = text.replace("‘", "'").replace("’", "'")
    text = text.replace("–", "-").replace("—", "-")
    text = re.sub(r"\s+", " ", text)
    return text.strip().lower()


def normalize_label(label):
    """Normalize labels so capitalization/spacing does not affect evaluation."""
    label = str(label).strip().lower()
    label = label.replace(" ", "_").replace("-", "_")
    return label


def get_sample_id(path):
    """Convert file names into the same sample ID.

    Examples:
    sample1_reference_blocks.json -> sample1
    sample1_llama_blocks.json     -> sample1
    sample1_reference__blocks.json -> sample1
    """
    stem = Path(path).stem.lower()
    stem = stem.replace("__", "_")

    suffixes = [
        "_reference_labels_blocks",
        "_reference_blocks",
        "_manual_labels_blocks",
        "_manual_blocks",
        "_llama_blocks",
        "_ai_blocks",
        "_detected_blocks",
        "_structured_blocks",
        "_blocks",
    ]

    for suffix in suffixes:
        if stem.endswith(suffix):
            stem = stem[:-len(suffix)]
            break

    stem = stem.replace("__", "_").strip("_")

    match = re.search(r"sample\s*[_-]*(\d+)", stem)
    if match:
        return f"sample{int(match.group(1))}"

    return stem


def load_json_blocks(path):
    """Load one JSON file and return clean block records."""
    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)

    # Support files where blocks are stored inside a dictionary.
    if isinstance(data, dict):
        for key in ["blocks", "structured_blocks", "data"]:
            if key in data and isinstance(data[key], list):
                data = data[key]
                break

    if not isinstance(data, list):
        raise ValueError(f"Expected list of blocks in {path}, but got {type(data)}")

    clean_blocks = []

    for i, block in enumerate(data):
        if not isinstance(block, dict):
            continue

        label = normalize_label(block.get("label", ""))
        text = str(block.get("text", "")).strip()

        if not text:
            continue

        clean_blocks.append({
            "block_index": i + 1,
            "label": label,
            "text": text,
            "norm_text": normalize_text(text),
        })

    return clean_blocks


def text_similarity(a, b):
    """Character-level text similarity. Range: 0 to 1."""
    a = normalize_text(a)
    b = normalize_text(b)

    if not a and not b:
        return 1.0
    if not a or not b:
        return 0.0

    return SequenceMatcher(None, a, b).ratio()


def token_set_similarity(a, b):
    """Word-overlap similarity. Useful when one block is split or merged."""
    a_tokens = set(normalize_text(a).split())
    b_tokens = set(normalize_text(b).split())

    if not a_tokens and not b_tokens:
        return 1.0
    if not a_tokens or not b_tokens:
        return 0.0

    return len(a_tokens & b_tokens) / len(a_tokens | b_tokens)


def label_match(reference_label, detected_label):
    return normalize_label(reference_label) == normalize_label(detected_label)


def relative_position(block_index, total_blocks):
    """Return normalized document position from 0 to 1."""
    if total_blocks <= 1:
        return 0.0
    return (block_index - 1) / (total_blocks - 1)


def order_difference_score(reference_index, detected_index, reference_total, detected_total):
    """Calculate how far the matched detected block is from expected document order.

    order_difference = 0 means perfect same relative position.
    order_score = 1 means perfect order match.
    """
    if detected_index is None:
        return None, None

    ref_pos = relative_position(reference_index, reference_total)
    det_pos = relative_position(detected_index, detected_total)
    order_difference = abs(ref_pos - det_pos)
    order_score = 1 - order_difference

    return round(order_difference, 4), round(order_score, 4)
```

## Step 6 — Match reference files with detected files

```python
reference_files = {
    get_sample_id(f): f
    for f in reference_dir.glob("*.json")
}

detected_files = {
    get_sample_id(f): f
    for f in detected_dir.glob("*.json")
}

matched_ids = sorted(
    set(reference_files.keys()) & set(detected_files.keys()),
    key=lambda x: int(re.search(r"\d+", x).group()) if re.search(r"\d+", x) else x
)

missing_detected = sorted(set(reference_files.keys()) - set(detected_files.keys()))
missing_reference = sorted(set(detected_files.keys()) - set(reference_files.keys()))

print("Number of reference files:", len(reference_files))
print("Number of detected files:", len(detected_files))
print("Number of matched file pairs:", len(matched_ids))
print("Matched IDs:", matched_ids)

if missing_detected:
    print("\nReference files without AI detected file:")
    print(missing_detected)

if missing_reference:
    print("\nAI detected files without reference file:")
    print(missing_reference)
```

## Step 7 — Compare blocks inside one sample with document order

```python
def compare_blocks_for_sample(
    sample_id,
    reference_path,
    detected_path,
    similarity_threshold=0.70,
    order_threshold=0.85,
):
    """Compare one reference JSON file with one detected JSON file.

    Evaluation logic:
    1. For each reference block, find the detected block with the best text similarity.
    2. Check whether the label is correct.
    3. Calculate document order difference using normalized block position.
    4. Assign status based on text, label, and order.
    """

    reference_blocks = load_json_blocks(reference_path)
    detected_blocks = load_json_blocks(detected_path)

    reference_total = len(reference_blocks)
    detected_total = len(detected_blocks)

    reference_rows = []
    used_detected_indices = set()

    for ref in reference_blocks:
        best = None
        best_score = -1
        best_seq_score = 0
        best_token_score = 0

        for det in detected_blocks:
            seq_score = text_similarity(ref["text"], det["text"])
            token_score = token_set_similarity(ref["text"], det["text"])

            # Text similarity remains the main matching criterion.
            combined_score = (seq_score * 0.60) + (token_score * 0.40)

            if combined_score > best_score:
                best_score = combined_score
                best = det
                best_seq_score = seq_score
                best_token_score = token_score

        if best is None:
            detected_index = None
            detected_label = "missing"
            detected_text = ""
            seq_score = 0
            token_score = 0
            combined_score = 0
            is_label_match = False
            order_diff = None
            order_score = None
            order_status = "missing"
            status = "missing"
        else:
            detected_index = best["block_index"]
            detected_label = best["label"]
            detected_text = best["text"]
            seq_score = best_seq_score
            token_score = best_token_score
            combined_score = best_score
            is_label_match = label_match(ref["label"], detected_label)

            order_diff, order_score = order_difference_score(
                reference_index=ref["block_index"],
                detected_index=detected_index,
                reference_total=reference_total,
                detected_total=detected_total,
            )

            if order_score is None:
                order_status = "missing"
            elif order_score >= order_threshold:
                order_status = "order_ok"
            else:
                order_status = "order_shifted"

            if combined_score < similarity_threshold:
                status = "missing_or_weak_match"
            elif is_label_match and order_status == "order_ok":
                status = "matched"
                used_detected_indices.add(detected_index)
            elif is_label_match and order_status == "order_shifted":
                status = "matched_but_order_shifted"
                used_detected_indices.add(detected_index)
            elif not is_label_match:
                status = "label_mismatch"
                used_detected_indices.add(detected_index)
            else:
                status = "needs_review"

        reference_rows.append({
            "sample_id": sample_id,
            "reference_file": reference_path.name,
            "detected_file": detected_path.name,
            "reference_block_index": ref["block_index"],
            "reference_total_blocks": reference_total,
            "reference_relative_position": relative_position(ref["block_index"], reference_total),
            "reference_label": ref["label"],
            "reference_text": ref["text"],
            "best_detected_block_index": detected_index,
            "detected_total_blocks": detected_total,
            "best_detected_relative_position": relative_position(detected_index, detected_total) if detected_index else None,
            "best_detected_label": detected_label,
            "best_detected_text": detected_text,
            "sequence_similarity": round(seq_score, 4),
            "token_set_similarity": round(token_score, 4),
            "combined_similarity": round(combined_score, 4),
            "label_match": is_label_match,
            "order_difference": order_diff,
            "order_score": order_score,
            "order_status": order_status,
            "status": status,
        })

    detected_rows = []

    for det in detected_blocks:
        if det["block_index"] not in used_detected_indices:
            detected_status = "extra_or_unmatched_ai_block"
        else:
            detected_status = "used_in_reference_match"

        detected_rows.append({
            "sample_id": sample_id,
            "detected_file": detected_path.name,
            "detected_block_index": det["block_index"],
            "detected_total_blocks": detected_total,
            "detected_relative_position": relative_position(det["block_index"], detected_total),
            "detected_label": det["label"],
            "detected_text": det["text"],
            "status": detected_status,
        })

    status_counts = Counter(row["status"] for row in reference_rows)
    label_counts_ref = Counter(block["label"] for block in reference_blocks)
    label_counts_det = Counter(block["label"] for block in detected_blocks)

    matched_count = status_counts.get("matched", 0)
    shifted_count = status_counts.get("matched_but_order_shifted", 0)
    label_mismatch_count = status_counts.get("label_mismatch", 0)
    weak_or_missing_count = status_counts.get("missing_or_weak_match", 0) + status_counts.get("missing", 0)

    order_values = [row["order_difference"] for row in reference_rows if row["order_difference"] is not None]
    average_order_difference = sum(order_values) / len(order_values) if order_values else None

    summary = {
        "sample_id": sample_id,
        "reference_file": reference_path.name,
        "detected_file": detected_path.name,
        "reference_block_count": reference_total,
        "detected_block_count": detected_total,
        "matched_count": matched_count,
        "matched_but_order_shifted_count": shifted_count,
        "label_mismatch_count": label_mismatch_count,
        "missing_or_weak_match_count": weak_or_missing_count,
        "extra_ai_block_count": sum(1 for row in detected_rows if row["status"] == "extra_or_unmatched_ai_block"),
        "match_rate": round(matched_count / reference_total, 4) if reference_total else 0,
        "text_label_match_rate_including_order_shifted": round((matched_count + shifted_count) / reference_total, 4) if reference_total else 0,
        "order_shift_rate": round(shifted_count / reference_total, 4) if reference_total else 0,
        "label_mismatch_rate": round(label_mismatch_count / reference_total, 4) if reference_total else 0,
        "average_order_difference": round(average_order_difference, 4) if average_order_difference is not None else None,
        "reference_label_counts": dict(label_counts_ref),
        "detected_label_counts": dict(label_counts_det),
    }

    return summary, reference_rows, detected_rows
```

## Step 8 — Run comparison for all matched samples

```python
all_summary_rows = []
all_reference_rows = []
all_detected_rows = []
full_json_report = {}

similarity_threshold = 0.70
order_threshold = 0.85

for sample_id in matched_ids:
    ref_path = reference_files[sample_id]
    det_path = detected_files[sample_id]

    summary, reference_rows, detected_rows = compare_blocks_for_sample(
        sample_id=sample_id,
        reference_path=ref_path,
        detected_path=det_path,
        similarity_threshold=similarity_threshold,
        order_threshold=order_threshold,
    )

    all_summary_rows.append(summary)
    all_reference_rows.extend(reference_rows)
    all_detected_rows.extend(detected_rows)

    full_json_report[sample_id] = {
        "summary": summary,
        "reference_block_analysis": reference_rows,
        "detected_block_analysis": detected_rows,
    }

summary_df = pd.DataFrame(all_summary_rows)
reference_df = pd.DataFrame(all_reference_rows)
detected_df = pd.DataFrame(all_detected_rows)

print("Summary rows:", len(summary_df))
print("Reference block rows:", len(reference_df))
print("Detected block rows:", len(detected_df))

summary_df
```

## Step 9 — View problematic blocks

```python
if len(reference_df) > 0:
    problems_df = reference_df[reference_df["status"] != "matched"].copy()
    print("Problematic reference blocks:", len(problems_df))
    display(problems_df[
        [
            "sample_id",
            "reference_block_index",
            "reference_label",
            "best_detected_block_index",
            "best_detected_label",
            "combined_similarity",
            "label_match",
            "order_difference",
            "order_score",
            "order_status",
            "status",
            "reference_text",
            "best_detected_text",
        ]
    ])
else:
    print("No reference comparison rows found.")
```

## Step 10 — View extra AI blocks

```python
if len(detected_df) > 0:
    extra_ai_df = detected_df[detected_df["status"] == "extra_or_unmatched_ai_block"].copy()
    print("Extra/unmatched AI blocks:", len(extra_ai_df))
    display(extra_ai_df[
        [
            "sample_id",
            "detected_block_index",
            "detected_label",
            "status",
            "detected_text",
        ]
    ])
else:
    print("No detected comparison rows found.")
```

## Step 11 — Per-label accuracy

```python
if len(reference_df) > 0:
    per_label_rows = []

    for label, group in reference_df.groupby("reference_label"):
        total = len(group)
        matched = (group["status"] == "matched").sum()
        matched_or_shifted = group["status"].isin(["matched", "matched_but_order_shifted"]).sum()
        label_mismatch = (group["status"] == "label_mismatch").sum()
        missing_or_weak = group["status"].isin(["missing", "missing_or_weak_match"]).sum()
        order_shifted = (group["status"] == "matched_but_order_shifted").sum()

        per_label_rows.append({
            "label": label,
            "reference_count": total,
            "strict_correct_count": matched,
            "strict_accuracy": round(matched / total, 4) if total else 0,
            "text_label_correct_including_order_shifted_count": matched_or_shifted,
            "text_label_accuracy_including_order_shifted": round(matched_or_shifted / total, 4) if total else 0,
            "label_mismatch_count": label_mismatch,
            "missing_or_weak_match_count": missing_or_weak,
            "order_shifted_count": order_shifted,
            "average_order_difference": round(group["order_difference"].dropna().mean(), 4),
            "average_similarity": round(group["combined_similarity"].mean(), 4),
        })

    per_label_accuracy_df = pd.DataFrame(per_label_rows).sort_values("label")
    display(per_label_accuracy_df)
else:
    per_label_accuracy_df = pd.DataFrame()
    print("No reference comparison rows found.")
```

## Step 12 — Confusion matrix

```python
if len(reference_df) > 0:
    confusion_input_df = reference_df.copy()

    # If text match is weak, count the prediction as missing instead of using the weak detected label.
    confusion_input_df["predicted_label_for_confusion"] = confusion_input_df.apply(
        lambda row: row["best_detected_label"] if row["combined_similarity"] >= similarity_threshold else "missing_or_weak_match",
        axis=1,
    )

    confusion_matrix_df = pd.crosstab(
        confusion_input_df["reference_label"],
        confusion_input_df["predicted_label_for_confusion"],
        rownames=["reference_label"],
        colnames=["detected_label"],
        dropna=False,
    )

    display(confusion_matrix_df)
else:
    confusion_matrix_df = pd.DataFrame()
    print("No reference comparison rows found.")
```

## Step 13 — Separate title / section / subsection / content performance

```python
target_labels = ["document_title", "section", "subsection", "content"]

if len(reference_df) > 0:
    separate_performance_df = per_label_accuracy_df[
        per_label_accuracy_df["label"].isin(target_labels)
    ].copy()

    # Keep the expected order of labels.
    separate_performance_df["label"] = pd.Categorical(
        separate_performance_df["label"],
        categories=target_labels,
        ordered=True,
    )
    separate_performance_df = separate_performance_df.sort_values("label")

    display(separate_performance_df)
else:
    separate_performance_df = pd.DataFrame()
    print("No reference comparison rows found.")
```

```python
summary_csv = output_dir / "summary_report.csv"
reference_csv = output_dir / "reference_block_level_report.csv"
detected_csv = output_dir / "detected_ai_block_level_report.csv"
per_label_csv = output_dir / "per_label_accuracy_report.csv"
confusion_csv = output_dir / "confusion_matrix_report.csv"
separate_performance_csv = output_dir / "title_section_subsection_content_performance.csv"
json_report_path = output_dir / "full_block_comparison_report.json"

summary_df.to_csv(summary_csv, index=False, encoding="utf-8-sig")
reference_df.to_csv(reference_csv, index=False, encoding="utf-8-sig")
detected_df.to_csv(detected_csv, index=False, encoding="utf-8-sig")
per_label_accuracy_df.to_csv(per_label_csv, index=False, encoding="utf-8-sig")
confusion_matrix_df.to_csv(confusion_csv, encoding="utf-8-sig")
separate_performance_df.to_csv(separate_performance_csv, index=False, encoding="utf-8-sig")

with open(json_report_path, "w", encoding="utf-8") as f:
    json.dump(full_json_report, f, indent=2, ensure_ascii=False)

print("Saved:")
print(summary_csv)
print(reference_csv)
print(detected_csv)
print(per_label_csv)
print(confusion_csv)
print(separate_performance_csv)
print(json_report_path)
```
